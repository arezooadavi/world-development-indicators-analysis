## Missing Values Analysis

```python
missing_Summary=pd.DataFrame({
    "Missing Count": df.isnull().sum(),
    "Missing Percentage": (df.isnull().mean()*100).round(2)      
    })
missing_Summary=missing_Summary.sort_values("Missing Percentage", ascending=False)
display(missing_Summary.head(20))
```

## Countries and Indicators

```python
print(f"Number of countries/regions: {df["Country Code"].nunique()}")
print(f"Number of indicators: {df["Indicator Code"].nunique()}")
print("\n Sample of indicators: ")
display(
    df[["Indicator Name","Indicator Code"]].drop_duplicates().head(20)
)
```

# **Development Indicators Selection**

```python
keywords=["GDP per capita","Life expectancy","School enrollment","Literacy","Poverty","Unemployment"]
for keyword in keywords:
    matches=df.loc[df["Indicator Name"].str.contains(keyword,case=False, na=False),
                   ["Indicator Name","Indicator Code"]].drop_duplicates()
    print(f"\n ---{keyword}---")
    display(matches.head(10))
```

## Selecting Key Development Indicators

```python
Indicator_keywords={"GDP per capita":"NY.GDP.PCAP.CD"
,"Life expectancy":"SP.DYN.LE00.IN"
,"School enrollment":"SE.SEC.ENRR"
,"Unemployment":"SL.UEM.TOTL.ZS"
}
for Name,code in Indicator_keywords.items():
    result=df.loc[df["Indicator Code"]==code, ["Indicator Name","Indicator Code"]].drop_duplicates()
    print(f"\n {Name}: ")
    display(result)
```

## Selecting the Analysis Year

```python
year_columns=[]
for col in df.columns:
    if str(col).isdigit():
        year_columns.append(col)
print(f"First available years: {year_columns[:10]}")
print(f"Last available years: {year_columns[-10:]}")
```

```python
selected_year="2020"
selected_code=list(Indicator_keywords.values())
analysis_df=df.loc[df["Indicator Code"].isin(selected_code),["Country Name","Country Code","Indicator Name","Indicator Code",selected_year]].copy()
analysis_df=analysis_df.rename(columns={selected_year: "Value"})
display(analysis_df.head(10)) 
```

```python
country_df=analysis_df.pivot_table(index=["Country Name","Country Code"],
                                   columns="Indicator Code",
                                   values="Value",
                                   aggfunc="mean").reset_index()
country_df.columns.name=None
display(country_df.head())
print(f"Final dataset shape: {country_df.shape}")
```

## Data Quality Assessment

```python
country_df.info()
display(country_df.describe())
display(country_df.isnull().sum().to_frame("missing Values"))
display(country_df.head())
```

## Missing Data Analysis

```python
missing_percent=(country_df.isnull().mean()*100).round(2)
display(missing_percent.to_frame("Missing Percetage").sort_values("Missing Percetage",ascending=False))
```

## Missing Data Visualization

```python
missing_percent.sort_values().plot(kind="bar",title="Missing Values by Development Indicator",
                                   ylabel="Missing Values (%)",xlabel="Indicator Code")
plt.xticks(rotation=45,ha="right")
plt.tight_layout()
plt.show()
```

## Correlation Analysis

```python
Indicator_columns=list(Indicator_keywords.values())
correlation_matrix=country_df[Indicator_columns].corr()
display(correlation_matrix)
plt.Figure(figsize=(9,6))
sns.heatmap(correlation_matrix,annot=True,cmap="coolwarm",fmt=".2f",vmin=-1,vmax=1)
plt.title("Correlation Between Development Indicators")
plt.tight_layout()
plt.show()
```

## GDP per Capita Analysis

```python
gdp_column="NY.GDP.PCAP.CD"
gdp_df=country_df[["Country Name","Country Code",gdp_column]].dropna()
gdp_df=gdp_df.sort_values(by=gdp_column,ascending=False)
display(gdp_df.head(10))
```

## Top 10 Country by per Capita

```python
top_gdp=gdp_df.head(10).sort_values(by=gdp_column,ascending=True)
plt.Figure(figsize=(12,7))
plt.barh(top_gdp["Country Name"],top_gdp[gdp_column])
plt.title("Top 10 Country by per Capita (2020)")
plt.xlabel("GDP per Capita (Current US$)")
plt.tight_layout()
plt.show()
```

## Life Expectancy Analysis

```python
life_column="SP.DYN.LE00.IN"
life_df=country_df[["Country Name","Country Code",life_column]].dropna()
life_df=life_df.sort_values(by=life_column,ascending= False)
display(life_df.head(10))
```

## Top 10 Countries by Life Expectancy

```python
top_life=life_df.head(10).sort_values(by=life_column,ascending=True)
plt.Figure(figsize=(12,7))
plt.barh(top_life["Country Name"],top_life[life_column])
plt.title("Top  10 Countries by Life Expectancy (2020)")
plt.xlabel("Life Expectancy at Birth (Years)")
plt.ylabel("Country")
plt.tight_layout()
plt.show
```

## GDP per Capita vs. life Expectancy

```python
comparison_df = country_df[["Country Name","Country Code",gdp_column,life_column]].dropna()
print(f"Number of observation : {len(comparison_df)}")
correlation = comparison_df[[gdp_column , life_column]].corr().iloc[0,1]
print(f"correlation: {round(correlation,3)}")
plt.Figure(figsize=(10,6))
sns.scatterplot(data=comparison_df,x=gdp_column,y=life_column)
plt.title("GDP per Capita vs. life Expectancy (2020)")
plt.xlabel("GDP per Capita (Current US$)")
plt.ylabel("life Expectancy (Years)")
plt.tight_layout()
plt.show()
```

## Education Indicator Anahysis

```python
education_column="SE.SEC.ENRR"
education_df=country_df[["Country Name", "Country Code",education_column]].dropna()
education_df=education_df.sort_values(by=education_column,ascending=False)
print(f"Number of countries/regions: {len(education_df)}")
display(education_df.head(10))
```

```python
top_education=education_df.head(10).sort_values(by=education_column,ascending=True)
plt .figure(figsize=(12,7))
plt.barh(top_education["Country Name"],
         top_education[education_column])
plt.title("Top 10 Countries by Secondary School Enrollment (2020)")
plt.xlabel("Enrollment Rate (%)")
plt.ylabel("Country")
plt.tight_layout()
plt.show()
```

## Unemployment Analysis

```python
unemployment_column="SL.UEM.TOTL.ZS"
unemployment_df=country_df[["Country Name","Country Code",unemployment_column]].dropna()
unemployment_df=unemployment_df.sort_values(by=unemployment_column,ascending=False)
display(unemployment_df.head(10))
```

```python
top_unemployment=unemployment_df.head(10).sort_values(by=unemployment_column,ascending=True)
plt .figure(figsize=(12,7))
plt.barh(top_unemployment["Country Name"],
         top_unemployment[unemployment_column])
plt.title("Top 10 Countries by Unemployment Rate (2020)")
plt.xlabel("unemployment Rate (%)")
plt.ylabel("Country")
plt.tight_layout()
plt.show()
```

## Development Indicators Correlation

```python
Indicator_columns=[gdp_column,life_column,education_column,unemployment_column]
correlation_data=country_df[Indicator_columns]
correlation_matrix=correlation_data.corr()
display(correlation_matrix)
plt.Figure(figsize=(10,7))
sns.heatmap(correlation_matrix,annot=True,cmap="coolwarm",fmt=" .2f",vmin=-1,vmax=1)
plt.title("Correlation Between Development Indicators (2020)")
plt.tight_layout()
plt.show()
```

## Summary Statistics

```python
summary_statistics=country_df[Indicator_columns].describe().T
summary_statistics["Missing Count"]=(country_df[Indicator_columns].isnull().sum())
summary_statistics["Missing Percentage"]=((country_df[Indicator_columns].isnull().mean()*100).round(2))
display(summary_statistics)
```

```python
country_df.to_csv("WDL_Development_Analysis.csv",index=False,encoding="utf _8_sig")
summary_statistics.to_csv("WDL_Statistics.csv",encoding="utf_8_sig")
print('analysis files saved successfully!')
```
